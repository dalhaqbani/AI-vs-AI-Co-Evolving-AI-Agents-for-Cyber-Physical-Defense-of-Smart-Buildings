##AI Engine Placeholder Logic Test


This notebook drafts and tests the Week 2 placeholder logic for `ai_engine.py`, before it gets copied into the actual backend file. It covers real threshold rules for **DHT22** (high temperature) and **servo lock** (unlocked state) — the two components with a meaningful single-reading check.

The remaining three components (PIR, fan/LED, push button) are intentionally left as placeholders. None of them have a meaningful single-reading threshold the way temperature or lock state do — motion, fan state, and button presses only become suspicious in context (timing, patterns, matching commands), which is real Layer 1/Layer 2 work for Sprint 3, not something a simple placeholder rule should try to fake.

```python
from dataclasses import dataclass
from typing import Optional

@dataclass
class AIDecision:
    status: str
    alert: Optional[str]
    action: Optional[str]
    score: float


def evaluate(component_type: str, data: dict) -> AIDecision:
    if component_type == "dht22":
        temp = data.get("temperature", 0)
        if temp > 35.0:
            return AIDecision(
                status="alert",
                alert=f"High temperature: {temp}°C",
                action="safe_mode",
                score=0.8,
            )

    elif component_type == "servo_lock":
        if data.get("state") == "unlocked":
            return AIDecision(
                status="alert",
                alert="Door reported unlocked",
                action="lock",
                score=0.8,
            )

    return AIDecision(status="normal", alert=None, action=None, score=0.0)
```

**Test** **Cell**

```python
def run_test(name, component_type, data, expected_status):
    result = evaluate(component_type, data)
    passed = result.status == expected_status
    print(f"{'PASS' if passed else ' FAIL'} — {name}")
    print(f"    Result: {result}")
    return passed

print("Running placeholder logic tests\n")
run_test("High temperature triggers alert", "dht22", {"temperature": 40.0}, "alert")
run_test("Normal temperature stays normal", "dht22", {"temperature": 22.0}, "normal")
run_test("Unlocked door triggers alert", "servo_lock", {"state": "unlocked"}, "alert")
run_test("Locked door stays normal", "servo_lock", {"state": "locked"}, "normal")
```
